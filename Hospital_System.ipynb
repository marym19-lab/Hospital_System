```python
#Name : Mariam Ayman Mohamed
class Patient:
  def __init__(self , patientId , name , age , gender , state) :
    self.patientId = patientId
    self.name = name
    self.age = age
    self.gender = gender
    self.state = state
  def patient_detail(self):
      print(f"patient id : {self.patientId} , Name : {self.name} , Age :{self.age} , State : {self.state}")


class Doctor:
  def __init__(self ,docId, name , age ,  specialization) :
    self.docId =docId
    self.name = name
    self.age = age
    self.specialization = specialization
  def doctor_detail(self):
    print(f' Doctor ID : {self.docId} , Name : {self.name} , Specialization : {self.specialization}')


class Appointment:
  def __init__(self , appointmentId , patient , doctor):
    self.appointmentId = appointmentId
    self.patient = patient
    self.doctor = doctor
  def appointment_detail(self):
    print(f' appointment ID : {self.appointmentId} , Patient name : {self.patient} , Doctor name : {self.doctor}')
#****************************************************************
class Hospital:
   def __init__(self):
     self.Patients=[]
     self.Doctors=[]
     self.Appointments=[]
     self.Appointment_counter=1
# ****************** patient **********************
   def add_patient(self):
     patientId=len(self.Patients)+1
     name = input("enter patient name : ")
     age = input("enter patient age : ")
     gender = input("enter patient gender : ")
     state = input("enter patient state : ")
     new_patient = Patient(patientId , name , age, gender , state)
     self.Patients.append(new_patient)
     print("sucess")
   def view_patient_info(self):
      if not self.Patients:
        print('no patients')
      else:
       for i in self.Patients:
         i.patient_detail()
   def search_Id(self):
      patient_ID=input("Enter Id : ")
      found=False
      for pID in self.Patients :
        if str(pID.patientId) == patient_ID :
         print('found')
         pID.patient_detail()
         found=True
         break
        else:
         print('Patient not found')
    # ************* Doctors *****************
   def add_doctors(self):
     doctorId = len(self.Doctors)+1
     name = input("enter doctor name : ")
     age = input("enter doctor age : ")
     specialization = input("enter doctor specialization : ")
     new_doctor = Doctor(doctorId , name , age, specialization)
     self.Doctors.append(new_doctor)
     print("sucess")

   def view_doctor_info(self):
      if not self.Doctors:
        print('No doctors')
      else:
       for i in self.Doctors:
         i.doctor_detail()
   def search_doctorID(self):
    docID = input("Enter Doctor ID: ")
    found = False
    for doctor in self.Doctors:
        if str(doctor.docId) == docID:
            print("Doctor found:")
            doctor.doctor_detail()
            found = True
            break
    else:
        print("Doctor not found")
        #*****************************************************
   def book_appointment(self) :
     appointmentId = len(self.Appointments)+1
     patient_name = input('Enter patient name : ')
     doctor_name = input('Enter doctor name : ')
     new_appointment = Appointment( appointmentId ,patient_name ,doctor_name )
     self.Appointments.append(new_appointment)
   def view_appointment_info(self):
      if not self.Appointments:
        print('No appointment')
      else:
       for i in self.Appointments:
         i.appointment_detail()


def show():
  print('---------------------------------')
  print('''
| Choise number for :                     |
|    1 : Add new patient                  |
|    2 : view patient info                |
|    3 : search by patient ID             |
|    4 : Add doctors                      |
|    5 : view doctor info                 |
|    6 : Search by doctor ID              |
|    7 : Book appointment                 |
|    8 : view appointment info            |
|    E : Exit                            |

''')
  print('-----------------------------------')
  h=Hospital()
  while True:

   c =input("Enter your choise : ")
   if c=='1':
     h.add_patient()
   elif c=='2':
     h.view_patient_info()
   elif c=='3':
     h.search_Id()
   elif c=='4':
     h.add_doctors()
   elif c=='5':
     h.view_doctor_info()
   elif c=='6':
     h.search_doctorID()
   elif c=='7':
     h.book_appointment()
   elif c=='8':
     h.view_appointment_info()
   elif c.upper()== 'E':
     break
show()
```
